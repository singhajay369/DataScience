# 📝 Topic: Mutable vs Immutable (Memory & Scope Logic)

### **1. The Core Mental Model (Analogy)**
* **Immutable (Stone Statue 🗿):** Change nahi kar sakte. Change karne ke liye **purana feko, naya banao**.

  Immutable modify karne ke liye global/nonlocal kyu chahiye?

    Because:
    ❗ Immutable cannot be modified — only REASSIGNED.

      Matlab:

          x = 5
        def f():
        x = x + 1   # ❌ new value ban rahi hai → rebinding → Python makes x local


  Assignment → makes x LOCAL, causing:

  Shadowing

  Or error (UnboundLocalError)

  To modify global/enclosing variables, you MUST tell Python:

  global x

  or nonlocal x

  Otherwise Python bolta hai:

  “You assigned to x → so x is LOCAL → global x untouched.”

    * *Ex:* `int`, `float`, `str`, `tuple`.




* **Mutable (School Bag 🎒):** Bag wahi rehta hai, bas andar ka **samaan badalta hai**.
    
    * mutable ko modify karne ke liye global/nonlocal nahi chahiye
      
      You are NOT changing the variable,
      
      you are only changing the object’s content.

    * *Ex:* `list`, `dict`, `set`.

          lst = [1, 2, 3]
          def f():
        lst.append(4) # allowed, no global needed
  
        lst.append() → list object modify hota hai
      
        But lst variable ka reference change nahi hota.
        
        Isliye Python ko koi issue nahi.

---

### **2. Memory Level Behavior (The "Why")**
* **Reassignment (`x = x + 1`):**
    * Python calculates `x + 1`.
    * Creates a **NEW object** at a **NEW memory address**.
    * Moves the label `x` to the new address. (Rebinding).
* **Mutation (`lst.append(4)`):**
    * Python goes to the **SAME memory address**.
    * Adds data inside the existing object directly. (In-place modification).

---

### **3. The LEGB/Scope Conflict (Why Global?)**
* **The Trap:** Python mein `=` (Assignment Operator) ka matlab hai **"Make a NEW Local Variable"**.

| Scenario | Code | What Python Thinks | Need Global? |
| :--- | :--- | :--- | :--- |
| **A. Reassignment (Immutable)** | `x = x + 1` | `=` dekha → "Oh, new local variable `x` bana rahe ho?" → Error if not global. | **YES** (`global x`) |
| **B. Mutation (Mutable)** | `lst.append(4)` | No `=` sign → "Oh, modify karna hai? Global `lst` use kar leta hu." | **NO** |

---

### **4. Summary Table (Cheat Sheet)**

| Type | Examples | Action | `=` Used? | Memory Address | Need `global`? |
| :--- | :--- | :--- | :--- | :--- | :--- |
| **Immutable** | `int`, `str`, `tuple` | `x = 5` | ✅ Yes | **Changes** (New Object) | **YES** |
| **Mutable** | `list`, `dict` | `.append()` | ❌ No | **Same** (In-place) | **NO** |

---

### **5. The Golden Rule (For Production Code)**
> 🔹 **Mutation (Internal Change)** $\rightarrow$ **NO Global needed.**
>
> 🔹 **Reassignment (Variable Change)** $\rightarrow$ **YES Global needed.**

---

# NAMESPACE vs SCOPE


 NAMESPACE → "Ghar"
 (Variable jahan rehta hai)

 SCOPE → "Awaaz ki range"
 (Variable jahan tak visible/accessible hota hai)


 -------------------------------
 Namespace Examples
 -------------------------------

x = 10     # Global namespace

def f():

y = 20   # Local namespace

print(x) # x is NOT local → Python checks outer namespaces


 -------------------------------
- x global namespace me store hota hai

- y local namespace me store hota hai

- f() ke andar x visible hai → kyunki global scope visible hota hai

- f() ke bahar y visible nahi hai → kyunki local scope limited hota hai


 -------------------------------
 FINAL DIFFERENCE (1-liners)
 -------------------------------
 Namespace = jahan variable STORE hota hai.
 Scope     = jahan variable ACCESSIBLE hota hai.

 Namespace → storage container
 Scope     → visibility rules (LEGB)

2. globals() → module ke global variables

3. inspect.getclosurevars() → closure ke variables

4. func.__closure__ → captured variables
