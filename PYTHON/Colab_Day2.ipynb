```python
for i in range (0, 3):
  pin = input("Enter your pin: ")
  if pin == "1234":
    print("Welcome")
  elif pin != "1234"

  while True:
    balance = 50000
    print(" 1. Check Balance","2. Withdraw Money", "3. Exit", sep = "\n")
    choice = input("Enter your choice: ")
    if choice == "1":
      print(f"Your balance is {balance}")
    elif choice == "2":
      amount = int(input("Enter amount to withdraw: "))
      if amount > balance:
        print("Insufficient balance, Garib aadmi! Paise nahi hain.")
      else:
        balance -= amount
      print(f"{amount} withdrawn successfully")
    elif choice == "3":
      print("Thank you for using our services!")
      break
    else:
      print("Invalid input")
  print("Wrong pin")
```
